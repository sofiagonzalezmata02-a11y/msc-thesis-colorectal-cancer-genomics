## Cell 1 — Create residue position column

```python
import re
import numpy as np

def extract_pos(protein_change: str):
    if protein_change is None or pd.isna(protein_change):
        return np.nan
    s = str(protein_change)

    # first run of digits (works for G12D, R1450*, H1490Yfs*18, etc.)
    m = re.search(r"(\d+)", s)
    return int(m.group(1)) if m else np.nan

mut["pos"] = mut["Protein_Change"].apply(extract_pos)

mut[["Protein_Change", "pos"]].head(15)
```

```python
import re

def mut_type_simple(s):
    s = str(s)
    if "fs" in s:
        return "frameshift"
    if "*" in s:
        return "nonsense"
    if re.match(r"^[A-Z]\d+[A-Z]$", s):
        return "missense"
    return "other"

mut["mut_type_simple"] = mut["Protein_Change"].apply(mut_type_simple)

mut["mut_type_simple"].value_counts()
```

```python
import pandas as pd

ALIGN_PATH = "clustalo-I20260210-232225-0362-86106448-p1m.fa"

def parse_fasta(text):
    recs = []
    h, seq = None, []
    for line in text.splitlines():
        line = line.strip()
        if not line:
            continue
        if line.startswith(">"):
            if h is not None:
                recs.append((h, "".join(seq)))
            h, seq = line, []
        else:
            seq.append(line)
    if h is not None:
        recs.append((h, "".join(seq)))
    return recs

aligned = parse_fasta(open(ALIGN_PATH).read())

# Find human aligned sequence
human_header, human_aln = None, None
for h, s in aligned:
    if "P01116" in h or "Homo sapiens" in h:
        human_header, human_aln = h, s
        break

assert human_aln is not None, "Human sequence (P01116 / Homo sapiens) not found in alignment."

# Map human residue positions (1..189) -> alignment columns
pos_to_col = {}
pos = 0
for col, aa in enumerate(human_aln):
    if aa != "-":
        pos += 1
        pos_to_col[pos] = col

def pct_identity_to_human(col_idx):
    human_aa = human_aln[col_idx]
    matches, total = 0, 0
    for _, seq in aligned:
        aa = seq[col_idx]
        if aa == "-":
            continue
        total += 1
        if aa == human_aa:
            matches += 1
    return (matches / total) if total else None, human_aa

rows = []
for p in range(1, 190):  # KRAS length = 189
    col = pos_to_col[p]
    score, human_aa = pct_identity_to_human(col)
    rows.append({"Gene": "KRAS", "pos": float(p), "human_aa": human_aa, "cons_score": score})

kras_cons = pd.DataFrame(rows)

# Sanity check hotspots
for p, expected in [(12, "G"), (13, "G"), (61, "Q")]:
    r = kras_cons.loc[kras_cons["pos"] == float(p)].iloc[0]
    print(f"Pos {p}: human_aa={r['human_aa']} (expected {expected}), cons={r['cons_score']:.3f}")

kras_cons.head()
```

## Step A1 — Setup (imports + helpers)

```python
import os, re, requests
from collections import OrderedDict

OUT_DIR = "uniprot_fastas"
os.makedirs(OUT_DIR, exist_ok=True)

def uniprot_search_first_accession(query, size=1):
    url = "https://rest.uniprot.org/uniprotkb/search"
    params = {"query": query, "format": "json", "fields": "accession", "size": size}
    r = requests.get(url, params=params)
    r.raise_for_status()
    data = r.json()
    if not data.get("results"):
        return None
    return data["results"][0]["primaryAccession"]

def uniprot_fetch_fasta_by_accession(acc):
    url = f"https://rest.uniprot.org/uniprotkb/{acc}.fasta"
    r = requests.get(url)
    r.raise_for_status()
    return r.text

def uniprot_stream_fasta(query, size=200):
    url = "https://rest.uniprot.org/uniprotkb/stream"
    params = {"query": query, "format": "fasta", "size": size}
    r = requests.get(url, params=params)
    if r.status_code != 200:
        print("Query failed:", query)
        print("Status:", r.status_code)
        print("Message:", r.text[:300])
        r.raise_for_status()
    return r.text

def parse_fasta(text):
    recs = []
    h, seq = None, []
    for line in text.splitlines():
        line = line.strip()
        if not line:
            continue
        if line.startswith(">"):
            if h is not None:
                recs.append((h, "".join(seq)))
            h, seq = line, []
        else:
            seq.append(line)
    if h is not None:
        recs.append((h, "".join(seq)))
    return recs

def get_species(header):
    m = re.search(r"OS=([^=]+?) OX=", header)
    return m.group(1).strip() if m else "Unknown"

def write_fasta(records, out_path):
    with open(out_path, "w") as f:
        for h, s in records:
            f.write(h + "\n")
            for i in range(0, len(s), 60):
                f.write(s[i:i+60] + "\n")
3
```

## Step A2 — Function: build an “orthologs FASTA” for one gene

```python
import os, re
from collections import OrderedDict

# Priority species taxon IDs
PRIORITY_TAXIDS = [
    10090,  # Mus musculus (mouse)
    10116,  # Rattus norvegicus (rat)
    9031,   # Gallus gallus (chicken)
    7955,   # Danio rerio (zebrafish)
    8355,   # Xenopus laevis
    9823,   # Sus scrofa (pig)
    9913,   # Bos taurus (cow)
]

def get_taxid(header):
    m = re.search(r"OX=(\d+)", header)
    return int(m.group(1)) if m else None

def get_species(header):
    m = re.search(r"OS=([^=]+?) OX=", header)
    return m.group(1).strip() if m else "Unknown"

def build_ortholog_fasta_for_gene_priority(gene, n_total=8, reviewed=False, length_tol=0.30):
    """
    Builds a FASTA with human + (priority species if available) + fallbacks up to n_total.
    length_tol=0.30 keeps sequences within ±30% of human length to avoid fragments/isoforms.
    """
    # 1) Find reviewed human UniProt entry for the gene
    human_q = f'(gene_exact:{gene}) AND (organism_id:9606) AND (reviewed:true)'
    human_acc = uniprot_search_first_accession(human_q)
    if human_acc is None:
        raise ValueError(f"Could not find reviewed human UniProt entry for {gene}")

    human_fasta = uniprot_fetch_fasta_by_accession(human_acc)
    human_records = parse_fasta(human_fasta)
    human_header, human_seq = human_records[0]
    human_len = len(human_seq)

    # 2) Fetch vertebrate sequences
    reviewed_part = " AND (reviewed:true)" if reviewed else ""
    ortho_q = f'(gene_exact:{gene}) AND (taxonomy_id:7742){reviewed_part}'
    ortho_fasta = uniprot_stream_fasta(ortho_q, size=800)
    ortho_records = parse_fasta(ortho_fasta)

    # 3) Deduplicate: keep ONE sequence per species (and skip human duplicates)
    seen_species = set(["Homo sapiens"])
    candidates = []  # (taxid, species, header, seq)

    for h, s in ortho_records:
        sp = get_species(h)
        if sp == "Homo sapiens":
            continue
        if sp in seen_species:
            continue

        # length filter to remove fragments / weird isoforms
        if human_len > 0:
            if not ((1 - length_tol) * human_len <= len(s) <= (1 + length_tol) * human_len):
                continue

        seen_species.add(sp)
        taxid = get_taxid(h)
        candidates.append((taxid, sp, h, s))

    # 4) Select in priority order, then fill with fallbacks
    selected = [(human_header, human_seq)]  # human always first

    # Make a lookup: taxid -> first candidate
    by_taxid = OrderedDict()
    for taxid, sp, h, s in candidates:
        if taxid is not None and taxid not in by_taxid:
            by_taxid[taxid] = (h, s, sp)

    # Add priority species if available
    for tx in PRIORITY_TAXIDS:
        if len(selected) >= n_total:
            break
        if tx in by_taxid:
            h, s, sp = by_taxid[tx]
            selected.append((h, s))

    # Fill remaining with other species (fallbacks), preferring closer length to human
    if len(selected) < n_total:
        remaining = []
        used_taxids = set([get_taxid(h) for h, _ in selected if get_taxid(h) is not None])

        for taxid, sp, h, s in candidates:
            if taxid in used_taxids:
                continue
            remaining.append((abs(len(s) - human_len), h, s, sp))

        remaining.sort(key=lambda x: x[0])  # closest length first

        for _, h, s, sp in remaining:
            if len(selected) >= n_total:
                break
            selected.append((h, s))

    # 5) Write FASTA
    out_path = os.path.join(OUT_DIR, f"{gene}_orthologs.fasta")
    write_fasta(selected, out_path)

    species_list = ["Homo sapiens"] + [get_species(h) for h, _ in selected[1:]]
    return {
        "gene": gene,
        "human_uniprot": human_acc,
        "n_sequences": len(selected),
        "out_path": out_path,
        "species": species_list
    }
```

## Cell 2 — Auto-rename based on the _HUMAN header

```python
import os, re

mnemonic_to_gene = {
    "P53_HUMAN": "TP53",
    "APC_HUMAN": "APC",
    "BRAF_HUMAN": "BRAF",
    "SMAD4_HUMAN": "SMAD4",
    "TGFR2_HUMAN": "TGFBR2",
    "PTEN_HUMAN": "PTEN",
    "PK3CA_HUMAN": "PIK3CA",
}

def find_human_mnemonic(fasta_path):
    with open(fasta_path) as f:
        for line in f:
            if line.startswith(">") and "_HUMAN" in line:
                m = re.search(r"\|([A-Z0-9]+_HUMAN)\b", line)
                if m:
                    return m.group(1)
    return None

# rename any .fa files that are not already named GENE_aligned.fa
for fn in os.listdir("."):
    if fn.endswith(".fa") and not fn.endswith("_aligned.fa"):
        mnemonic = find_human_mnemonic(fn)
        if mnemonic in mnemonic_to_gene:
            gene = mnemonic_to_gene[mnemonic]
            new_name = f"{gene}_aligned.fa"
            print(f"Renaming: {fn}  -->  {new_name}")
            os.rename(fn, new_name)

print("\nAligned files now:")
!ls -la *aligned.fa
```

## Step B1 — Conservation from aligned FASTA (percent identity to human residue)

```python
import pandas as pd
import numpy as np

def conservation_table_from_alignment(aligned_fa, gene):
    aligned = parse_fasta(open(aligned_fa).read())

    # get human aligned sequence
    human_header, human_aln = None, None
    for h, s in aligned:
        if "_HUMAN" in h:
            human_header, human_aln = h, s
            break
    assert human_aln is not None, f"Human sequence not found in {aligned_fa}"

    # other aligned sequences
    other_alns = [s for h, s in aligned if h != human_header]
    L = len(human_aln)

    rows = []
    human_pos = 0  # counts residues in human (ignores gaps)

    for i in range(L):
        h_aa = human_aln[i]
        if h_aa == "-":
            continue  # no human residue here -> no human position

        human_pos += 1
        # compare other species at this alignment column
        col = [seq[i] for seq in other_alns]
        valid = [aa for aa in col if aa != "-"]  # ignore gaps in others

        if len(valid) == 0:
            cons = np.nan
        else:
            cons = sum(aa == h_aa for aa in valid) / len(valid)

        rows.append({"Gene": gene, "pos": float(human_pos), "human_aa": h_aa, "cons_score": cons})

    return pd.DataFrame(rows)
```

## Step B2 — Run it for all 7 genes (and later you can add KRAS)

## Add Gene as covariate (gene-adjusted logistic model)

```python
import pandas as pd
import statsmodels.api as sm

# 1) Make gene dummies (0/1)
gene_dummies = pd.get_dummies(residue_tbl["Gene"], drop_first=True)

# 2) Build design matrix
X_gene = pd.concat([residue_tbl[["cons_score"]], gene_dummies], axis=1)

# 3) Force numeric (important!)
X_gene = X_gene.apply(pd.to_numeric, errors="coerce")

# 4) Add intercept
X_gene = sm.add_constant(X_gene)

# 5) Outcome as int
y = residue_tbl["recurrent_binary"].astype(int)

# 6) Drop any rows that became NaN during conversion (just in case)
mask = X_gene.notna().all(axis=1) & y.notna()
X_gene2 = X_gene.loc[mask]
y2 = y.loc[mask]

print("Design matrix dtypes:\n", X_gene2.dtypes)
print("Rows used:", len(y2), " / ", len(y))

model_gene = sm.Logit(y2, X_gene2).fit(disp=False)
print(model_gene.summary())
```
